```python
import seaborn as sns
import matplotlib.pyplot as plt
print(sns.get_dataset_names())
```

```python
tips= sns.load_dataset("tips")
tips.head()
```

```python
tips.info()
```

1) LINE PLOT

```python
tips = sns.load_dataset('tips')
sns.lineplot(x="total_bill", y="tip", data=tips)
plt.title('Line Plot')
plt.grid(True)
plt.show()
```

2) SCATTER PLOT

```python
tips = sns.load_dataset('tips')
sns.scatterplot(x="total_bill", y="tip", data=tips)
plt.title('Scatter Plot')
plt.grid(True)
plt.show()
```

3) BAR PLOT

```python
tips = sns.load_dataset('tips')
sns.barplot(x="total_bill", y="tip", data=tips)
plt.title('Bar Plot')
plt.xlabel("x")
plt.ylabel("y")
# plt.grid(True)
plt.show()
```

4) BOX PLOT

```python
tips = sns.load_dataset('tips')
sns.boxplot(x="total_bill", y="tip", data=tips)
plt.title('Box Plot')
plt.xlabel("x")
plt.ylabel("y")
# plt.grid(True)
plt.show()
```

5) HEATMAP IN SEABORN

```python
flights = sns.load_dataset("flights")
flights
```

```python
pv = flights.pivot(index="month", columns="year", values="passengers")
pv
```

```python
sns.heatmap(pv, annot=True, fmt="d", cmap="YlGnBu")
```
